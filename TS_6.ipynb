## Introducción y objetivos

$\quad$ En este trabajo analizamos el comportamiento en frecuencia de tres circuitos pasivos RLC de segundo orden (a, b y c). Aunque usan los mismos componentes, cambia cómo están conectados y entre qué puntos se toma la salida, y eso hace que cada uno filtre la señal de forma distinta: atenuando las altas frecuencias, las bajas, o dejando pasar solo una banda alrededor de la resonancia.

$\quad$ El objetivo es comparar la selectividad de cada configuración a partir de su módulo y su fase, y ver cómo varían las pendientes de atenuación, los desfasajes y el ancho de banda.

$\quad$ Para esto seguimos tres pasos, con los valores normalizados de la consigna ($\omega_0 = 1\text{ rad/s}$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$):

1. **Función de transferencia:** deducir $H(s) = \frac{V_2(s)}{V_1(s)}$ de cada circuito.
2. **Respuesta en frecuencia y diagramas de Bode:** evaluar $H(s)$ en $s = j\omega$ para obtener el módulo $|H(j\omega)|$ (lineal y en dB) y la fase $\angle H(j\omega)$. La fase se analiza para $\omega \to 0$, $\omega = \omega_0$ y $\omega \to \infty$, y se trazan las asíntotas de bajas y altas frecuencias.
3. **Clasificación del filtro y ancho de banda:** identificar si cada circuito es pasa-bajos, pasa-altos o pasa-banda, y calcular su ancho de banda ($BW$) con los puntos de $-3\text{ dB}$.

## Circuito a) - Malla RLC Serie con salida en Capacitor C



### 1. Deducción Analítica de la Función de Transferencia $H(s)$

$\quad$ El circuito a) corresponde a una malla serie $RLC$ donde la tensión de entrada $V_i(s)$ se aplica a los terminales de entrada y la tensión de salida $V_o(s)$ se toma sobre los bornes del capacitor $C$ en circuito abierto ($I_2 = 0$).

Definiendo las impedancias en el dominio de Laplace:
* $Z_L(s) = sL$
* $Z_R(s) = R$
* $Z_C(s) = \frac{1}{sC}$

Por divisor de tensión en la malla:
$$V_o(s) = V_i(s) \cdot \frac{Z_{\text{salida}}}{Z_{\text{total}}} = V_i(s) \cdot \frac{Z_C}{Z_R + Z_C + Z_L}$$

$$H(s) = \frac{V_o(s)}{V_i(s)} = \frac{\frac{1}{sC}}{R + \frac{1}{sC} + sL}$$

Multiplicando numerador y denominador por $sC$:
$$H(s) = \frac{1}{sCR + 1 + s^2CL} = \frac{1}{s^2CL + sCR + 1}$$

Dividiendo numerador y denominador por el coeficiente principal $CL$:
$$H(s) = \frac{\frac{1}{CL}}{s^2 + \frac{R}{L}s + \frac{1}{CL}}$$

### Determinación de los Componentes Normalizados
Se usan las condiciones de diseño de la consigna:
$$\omega_0 = 1\text{ rad/s}, \quad R = 1\,\Omega, \quad Q = \frac{\sqrt{2}}{2}$$

Comparando el denominador con la forma canónica de segundo orden $D(s) = s^2 + \frac{\omega_0}{Q}s + \omega_0^2$:

1. **Frecuencia natural $\omega_0$:**
   $$\omega_0^2 = \frac{1}{LC} = 1 \implies LC = 1 \implies L = \frac{1}{C}$$

2. **Factor de calidad $Q$:**
   $$\frac{\omega_0}{Q} = \frac{R}{L} \implies Q = \frac{\omega_0 L}{R}$$
   Reemplazando $\omega_0 = 1$ y $R = 1\,\Omega$:
   $$Q = \frac{1 \cdot L}{1} = L \implies L = Q = \frac{\sqrt{2}}{2}\text{ H}$$

3. **Capacitancia $C$:**
   $$C = \frac{1}{L} = \frac{1}{\frac{\sqrt{2}}{2}} = \frac{2}{\sqrt{2}} = \sqrt{2}\text{ F}$$

Verificamos los coeficientes del polinomio:
* Término independiente: $\frac{1}{CL} = \frac{1}{\sqrt{2} \cdot \frac{\sqrt{2}}{2}} = \frac{1}{1} = 1$
* Coeficiente lineal: $\frac{R}{L} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}$

Sustituyendo estos valores numéricos, la función de transferencia normalizada resulta:

$$\boxed{H(s) = \frac{1}{s^2 + \sqrt{2}s + 1}}$$

### 2. Respuesta en Frecuencia y Análisis Asintótico

Partiendo de la evaluación de la transferencia mediante $s = j\omega$ (donde $s^2 = -\omega^2$):
$$H(j\omega) = \frac{1}{1 - \omega^2 + j\sqrt{2}\omega} = \frac{1}{(1 - \omega^2) + j\sqrt{2}\omega}$$

#### A. Módulo Analítico $|H(j\omega)|$ y Ganancia en dB

Para hallar el módulo, calculamos el módulo del denominador:
$$|H(j\omega)| = \frac{1}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

Desarrollando el binomio y el término imaginario bajo la raíz:
$$|H(j\omega)| = \frac{1}{\sqrt{(1 - 2\omega^2 + \omega^4) + 2\omega^2}} = \frac{1}{\sqrt{1 + \omega^4}}$$

#### Expresión en Decibeles
$$|H(j\omega)|_{\text{dB}} = 20 \log_{10} |H(j\omega)| = 20 \log_{10} \left( \frac{1}{\sqrt{1 + \omega^4}} \right)$$

Aplicando propiedades de los logaritmos ($\log(x^{-1/2}) = -\frac{1}{2}\log(x)$):
$$|H(j\omega)|_{\text{dB}} = -10 \log_{10}(1 + \omega^4)$$

#### Análisis Asintótico del Módulo:
* **Bajas Frecuencias ($\omega \to 0$, $\omega \ll 1$):**
  $$\omega^4 \ll 1 \implies 1 + \omega^4 \approx 1$$
  $$|H(j\omega)| \approx 1 \implies |H(j\omega)|_{\text{dB}} \approx -10 \log_{10}(1) = 0 \text{ dB}$$
  *(Asíntota horizontal de baja frecuencia en $0 \text{ dB}$).*

* **Altas Frecuencias ($\omega \to \infty$, $\omega \gg 1$):**
  $$\omega^4 \gg 1 \implies 1 + \omega^4 \approx \omega^4$$
  $$|H(j\omega)| \approx \frac{1}{\sqrt{\omega^4}} = \frac{1}{\omega^2}$$
  $$|H(j\omega)|_{\text{dB}} \approx -10 \log_{10}(\omega^4) = -40 \log_{10}(\omega)$$
  *(Asíntota con pendiente de $-40 \text{ dB/década}$).*

* **Frecuencia de Corte Característica ($\omega = \omega_0 = 1 \text{ rad/s}$):**
  $$|H(j1)| = \frac{1}{\sqrt{1 + 1^4}} = \frac{1}{\sqrt{2}} \approx 0{,}7071$$
  $$|H(j1)|_{\text{dB}} = -10 \log_{10}(1 + 1) = -10 \log_{10}(2) \approx -3{,}01 \text{ dB}$$
  *(La ganancia cae exactamente $3 \text{ dB}$ respecto a la banda de paso).*


#### B. Fase Analítica $\phi(\omega)$

La fase de la función de transferencia se calcula mediante el argumento del cociente:
$$\phi(\omega) = \arg\{H(j\omega)\} = \arg\{1\} - \arg\{(1 - \omega^2) + j\sqrt{2}\omega\}$$
$$\phi(\omega) = 0 - \arctan\left( \frac{\sqrt{2}\omega}{1 - \omega^2} \right) = -\arctan\left( \frac{\sqrt{2}\omega}{1 - \omega^2} \right)$$

#### Análisis Asintótico y Puntos Notables de la Fase:
* **Bajas Frecuencias ($\omega \to 0$):**
  $$\phi(0) = -\arctan(0) = 0 \text{ rad} = 0^\circ$$

* **Frecuencia de Corte ($\omega = 1$):**
  Cuando $\omega \to 1^-$, el denominador tiende a $0^+$ en su parte real:
  $$\phi(1) = -\arctan\left( \frac{\sqrt{2}}{0^+} \right) = -\frac{\pi}{2} \text{ rad} = -90^\circ$$
  *(El denominador se encuentra exactamente sobre el eje imaginario positivo $(0, \sqrt{2})$).*

* **Altas Frecuencias ($\omega \to \infty$):**
  Para $\omega > 1$, la parte real es negativa ($1 - \omega^2 < 0$) y la parte imaginaria es positiva ($\sqrt{2}\omega > 0$). Por lo tanto, el número complejo en el denominador pertenece al segundo cuadrante.
  Tomando el límite cuando $\omega \to \infty$, el ángulo del denominador tiende a $\pi$:
  $$\phi(\infty) = 0 - \pi = -\pi \text{ rad} = -180^\circ$$


### 3. Clasificación y Ancho de Banda

Se trata de un **Filtro Pasa-Bajos (Low-Pass Filter) de 2.º orden** con aproximación Butterworth ($Q = \frac{\sqrt{2}}{2} \approx 0{,}707$), caracterizado por dejar pasar la componente continua y las bajas frecuencias con ganancia unitaria ($0\text{ dB}$) y presentar una atenuación asintótica de $-40\text{ dB/década}$ ($-12\text{ dB/octava}$) en la banda de rechazo.

**Frecuencia de Corte ($\omega_c$):**
$$\omega_c = 1\text{ rad/s} \quad \left(f_c = \frac{\omega_c}{2\pi} \approx 0{,}159\text{ Hz}\right)$$

En este punto la ganancia cae exactamente $-3\text{ dB}$ (módulo lineal igual a $0{,}707$) y la fase presenta un retraso de $-90^\circ$.

**Análisis de la Respuesta y Puntos Clave:**
* **En bajas frecuencias ($\omega \to 0$):** El inductor actúa como cortocircuito a masa ($Z_L \to 0$) y el capacitor como circuito abierto ($Z_C \to \infty$), por lo que toda la tensión de entrada cae sobre la salida ($0\text{ dB}$, ganancia lineal 1).
* **En la frecuencia de corte ($\omega_c = 1\text{ rad/s}$):** La ganancia cae a $-3\text{ dB}$, marcando el límite superior de la banda de paso.
* **En altas frecuencias ($\omega \to \infty$):** El capacitor cortocircuita la salida a masa ($Z_C \to 0$) y el inductor bloquea el paso de corriente, logrando una atenuación rápida de $-40\text{ dB/década}$ y un retraso de fase que tiende asintóticamente a $-180^\circ$.

El ancho de banda del filtro queda definido como:
$$
\text{BW} = [0, \omega_c] = [0, 1]\text{ rad/s}
$$


## Circuito b) - Red RC Paralelo en serie con Inductor L a tierra



### 1. Deducción Analítica de la Función de Transferencia $H(s)$

$\quad$ En esta topología, la rama serie superior está compuesta por la conexión en paralelo de una resistencia $R$ y un capacitor $C$, mientras que la rama de derivación a tierra consiste en un inductor $L$. La tensión de salida $V_o(s)$ se toma sobre los bornes del inductor $L$.

Impedancia de la rama serie ($Z_1$):
$$Z_1(s) = Z_C \parallel Z_R = \frac{Z_C \cdot Z_R}{Z_C + Z_R} = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{sCR + 1}$$

Impedancia de la rama de derivación a masa ($Z_2$):
$$Z_2(s) = Z_L(s) = sL$$

Por divisor de tensión:
$$V_o(s) = V_i(s) \cdot \frac{Z_2}{Z_1 + Z_2} = V_i(s) \cdot \frac{sL}{sL + \frac{R}{sCR + 1}}$$

Multiplicando numerador y denominador por $(sCR + 1)$:
$$H(s) = \frac{V_o(s)}{V_i(s)} = \frac{sL(sCR + 1)}{sL(sCR + 1) + R} = \frac{s^2 LCR + sL}{s^2 LCR + sL + R}$$

Dividiendo numerador y denominador por el coeficiente principal $LCR$:
$$H(s) = \frac{s^2 + \frac{sL}{LCR}}{s^2 + \frac{sL}{LCR} + \frac{R}{LCR}} = \frac{s^2 + \frac{s}{CR}}{s^2 + \frac{s}{CR} + \frac{1}{LC}}$$

### Determinación de los Componentes Normalizados
Imponiendo las especificaciones de diseño:
$$\omega_0 = 1\text{ rad/s}, \quad R = 1\,\Omega, \quad Q = \frac{\sqrt{2}}{2}$$

Comparando el denominador con la forma canónica de segundo orden $D(s) = s^2 + \frac{\omega_0}{Q}s + \omega_0^2$:

1. **Frecuencia natural $\omega_0$:**
   $$\omega_0^2 = \frac{1}{LC} = 1 \implies L = \frac{1}{C}$$

2. **Factor de calidad $Q$:**
   $$\frac{\omega_0}{Q} = \frac{1}{CR} \implies \frac{1}{\frac{\sqrt{2}}{2}} = \frac{1}{C \cdot 1} \implies \sqrt{2} = \frac{1}{C} \implies C = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2}\text{ F}$$

3. **Inductancia $L$:**
   $$L = \frac{1}{C} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}\text{ H}$$

Verificamos los coeficientes:
* Término lineal: $\frac{1}{CR} = \frac{1}{\frac{\sqrt{2}}{2} \cdot 1} = \sqrt{2}$
* Término independiente: $\frac{1}{LC} = \frac{1}{\sqrt{2} \cdot \frac{\sqrt{2}}{2}} = 1$
* Numerador: $s^2 + \frac{s}{CR} = s^2 + \sqrt{2}s = s(s + \sqrt{2})$

Reemplazando estos valores, la función de transferencia normalizada resulta:

$$\boxed{H(s) = \frac{s^2 + \sqrt{2}s}{s^2 + \sqrt{2}s + 1} = \frac{s(s + \sqrt{2})}{s^2 + \sqrt{2}s + 1}}$$

### 2. Respuesta en Frecuencia y Análisis Asintótico 

Evaluando en $s = j\omega$ (con $s^2 = -\omega^2$):

$$H(j\omega) = \frac{-\omega^2 + j\omega\sqrt{2}}{-\omega^2 + j\omega\sqrt{2} + 1} = \frac{-\omega^2 + j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$

#### A. Módulo Analítico $|H(j\omega)|$ y Ganancia en dB

$$|H(j\omega)| = \frac{\sqrt{(-\omega^2)^2 + (\sqrt{2}\omega)^2}}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}} = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{1 + \omega^4}} = \frac{\omega \sqrt{\omega^2 + 2}}{\sqrt{1 + \omega^4}}$$

En decibeles:
$$|H(j\omega)|_{\text{dB}} = 20 \log_{10} \left( \omega \sqrt{\omega^2 + 2} \right) - 10 \log_{10}(1 + \omega^4)$$

#### Análisis Asintótico de Módulo:
* **Bajas frecuencias ($\omega \to 0$, $\omega \ll 1$):**
  $$\omega^2 + 2 \approx 2, \quad 1 + \omega^4 \approx 1$$
  $$|H(j\omega)| \approx \frac{\omega \sqrt{2}}{1} = \sqrt{2}\omega$$
  $$|H(j\omega)|_{\text{dB}} \approx 20 \log_{10}(\sqrt{2}\omega) = 20 \log_{10}(\sqrt{2}) + 20 \log_{10}(\omega) \approx 3{,}01 + 20 \log_{10}(\omega)$$
  *(Recta asíntota de baja frecuencia con pendiente de $+20\text{ dB/década}$ debido al cero en el origen $s=0$).*
  *(Notar que la asíntota cruza $0\text{ dB}$ en $\omega = \frac{1}{\sqrt{2}} \approx 0{,}707\text{ rad/s}$).*

* **Altas frecuencias ($\omega \to \infty$, $\omega \gg 1$):**
  $$\omega^2 + 2 \approx \omega^2, \quad 1 + \omega^4 \approx \omega^4$$
  $$|H(j\omega)| \approx \frac{\omega \cdot \omega}{\omega^2} = 1$$
  $$|H(j\omega)|_{\text{dB}} \approx 20 \log_{10}(1) = 0\text{ dB}$$
  *(Asíntota horizontal de alta frecuencia en $0\text{ dB}$).*

* **Frecuencia característica ($\omega = \omega_0 = 1\text{ rad/s}$):**
  $$|H(j1)| = \frac{1 \cdot \sqrt{1^2 + 2}}{\sqrt{1 + 1^4}} = \frac{\sqrt{3}}{\sqrt{2}} = \sqrt{1{,}5} \approx 1{,}2247$$
  $$|H(j1)|_{\text{dB}} = 20 \log_{10}(1{,}2247) \approx +1{,}76\text{ dB}$$
  *(Presenta un **sobrepico de $+1{,}76\text{ dB}$** en torno a $\omega = 1\text{ rad/s}$, generado por la presencia del cero real en $s = -\sqrt{2}$).*


#### B. Fase Analítica $\phi(\omega)$

$$\phi(\omega) = \arg\{-\omega^2 + j\sqrt{2}\omega\} - \arg\{(1 - \omega^2) + j\sqrt{2}\omega\}$$

Analizando el ángulo del numerador $N(j\omega) = -\omega^2 + j\sqrt{2}\omega$:
* Parte real $= -\omega^2 < 0$ (para todo $\omega > 0$).
* Parte imaginaria $= \sqrt{2}\omega > 0$.
* Por ende, el numerador se ubica siempre en el **segundo cuadrante**:
  $$\arg\{N(j\omega)\} = \pi - \arctan\left( \frac{\sqrt{2}\omega}{\omega^2} \right) = \pi - \arctan\left( \frac{\sqrt{2}}{\omega} \right)$$
  O factorizando $j\omega$:
  $$N(j\omega) = j\omega (\sqrt{2} + j\omega) \implies \arg\{N(j\omega)\} = \frac{\pi}{2} + \arctan\left( \frac{\omega}{\sqrt{2}} \right)$$

Analizando el ángulo del denominador $D(j\omega) = (1 - \omega^2) + j\sqrt{2}\omega$:
* Si $\omega < 1$: primer cuadrante $\implies \arg\{D\} = \arctan\left( \frac{\sqrt{2}\omega}{1 - \omega^2} \right)$
* Si $\omega = 1$: semieje imaginario positivo $\implies \arg\{D\} = \frac{\pi}{2}$
* Si $\omega > 1$: segundo cuadrante $\implies \arg\{D\} = \pi - \arctan\left( \frac{\sqrt{2}\omega}{\omega^2 - 1} \right)$

#### Evaluación de Puntos Notables de Fase:
* **$\omega \to 0$:**
  $$\arg\{N\} \to \frac{\pi}{2} + \arctan(0) = \frac{\pi}{2}, \quad \arg\{D\} \to \arctan(0) = 0$$
  $$\phi(0) = \frac{\pi}{2} - 0 = \frac{\pi}{2}\text{ rad} = +90^\circ$$

* **$\omega = 1$:**
  $$\arg\{N\} = \pi - \arctan(\sqrt{2}) \approx \pi - 0{,}9553\text{ rad} \approx 2{,}1863\text{ rad} \approx 125{,}26^\circ$$
  $$\arg\{D\} = \frac{\pi}{2} \approx 1{,}5708\text{ rad} = 90^\circ$$
  $$\phi(1) = 2{,}1863 - 1{,}5708 = 0{,}6155\text{ rad} \approx 35{,}26^\circ$$

* **$\omega \to \infty$:**
  $$\arg\{N\} \to \pi - \arctan(0) = \pi, \quad \arg\{D\} \to \pi - \arctan(0) = \pi$$
  $$\phi(\infty) = \pi - \pi = 0\text{ rad} = 0^\circ$$


### 3. Clasificación y Ancho de Banda

Observando el diagrama de Bode y los puntos característicos, confirmamos que el circuito actúa como un filtro pasa-altos de 2.º orden. Debido a que en bajas frecuencias ($\omega \to 0$) la ganancia es nula, arrancando desde $-\infty\text{ dB}$ (módulo lineal igual a $0$). El circuito bloquea por completo la componente continua y las frecuencias bajas. En el gráfico, la curva sube rápidamente hacia la banda de paso con una pendiente de $+40\text{ dB/década}$. En la frecuencia de corte ($\omega_c = 1\text{ rad/s}$) al llegar a $\omega = 1\text{ rad/s}$, la ganancia alcanza los $-3\text{ dB}$ (módulo $0{,}707$). Este punto marca el límite inferior a partir del cual el filtro empieza a dejar pasar la señal. En altas frecuencias ($\omega \to \infty$) para frecuencias elevadas, la ganancia se estabiliza en $0\text{ dB}$ (módulo lineal igual a $1$). Todas las componentes de alta frecuencia se transmiten a la salida limpias y sin sufrir atenuación.

El ancho de banda del filtro queda definido como:

$$
\text{BW} = [\omega_c,\infty) = [1,\infty)\text{ rad/s}
$$

Por lo tanto, el filtro deja pasar las componentes frecuenciales superiores a la frecuencia de corte $\omega_c = 1\text{ rad/s}$.


## Circuito c) - Resistor R en serie con Tanque LC Paralelo a tierra



### 1. Deducción Analítica de la Función de Transferencia $H(s)$

$\quad$ En esta topología, la rama serie superior está compuesta por una resistencia $R$, mientras que la rama de derivación a tierra está formada por la conexión en paralelo de un inductor $L$ y un capacitor $C$ (tanque resonante $LC$). La tensión de salida $V_o(s)$ se toma sobre este paralelo.

Impedancia del tanque resonante en paralelo ($Z_2$):
$$Z_2(s) = Z_L \parallel Z_C = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{\frac{sL}{sC}}{\frac{s^2 LC + 1}{sC}} = \frac{sL}{s^2 LC + 1}$$

Impedancia de la rama serie ($Z_1$):
$$Z_1(s) = R$$

Por divisor de tensión:
$$V_o(s) = V_i(s) \cdot \frac{Z_2}{Z_1 + Z_2} = V_i(s) \cdot \frac{\frac{sL}{s^2 LC + 1}}{R + \frac{sL}{s^2 LC + 1}}$$

Multiplicando numerador y denominador por $(s^2 LC + 1)$:
$$H(s) = \frac{V_o(s)}{V_i(s)} = \frac{sL}{R(s^2 LC + 1) + sL} = \frac{sL}{s^2 LCR + sL + R}$$

Dividiendo numerador y denominador por el coeficiente principal $LCR$:
$$H(s) = \frac{\frac{sL}{LCR}}{s^2 + \frac{sL}{LCR} + \frac{R}{LCR}} = \frac{\frac{s}{CR}}{s^2 + \frac{s}{CR} + \frac{1}{LC}}$$

### Determinación de los Componentes Normalizados
Imponiendo las especificaciones de diseño:
$$\omega_0 = 1\text{ rad/s}, \quad R = 1\,\Omega, \quad Q = \frac{\sqrt{2}}{2}$$

Comparando el denominador con la forma canónica de segundo orden $D(s) = s^2 + \frac{\omega_0}{Q}s + \omega_0^2$:

1. **Frecuencia natural $\omega_0$:**
   $$\omega_0^2 = \frac{1}{LC} = 1 \implies LC = 1 \implies L = \frac{1}{C}$$

2. **Factor de calidad $Q$:**
   $$\frac{\omega_0}{Q} = \frac{1}{CR} \implies \frac{1}{\frac{\sqrt{2}}{2}} = \frac{1}{C \cdot 1} \implies \sqrt{2} = \frac{1}{C} \implies C = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2}\text{ F}$$

3. **Inductancia $L$:**
   $$L = \frac{1}{C} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}\text{ H}$$

Verificamos los coeficientes:
* Término lineal: $\frac{1}{CR} = \frac{1}{\frac{\sqrt{2}}{2} \cdot 1} = \sqrt{2}$
* Término independiente: $\frac{1}{LC} = \frac{1}{\sqrt{2} \cdot \frac{\sqrt{2}}{2}} = 1$
* Numerador: $\frac{s}{CR} = \sqrt{2}s$

Sustituyendo estos valores numéricos, la función de transferencia normalizada resulta:

$$\boxed{H(s) = \frac{\sqrt{2}s}{s^2 + \sqrt{2}s + 1}}$$

### 2. Respuesta en Frecuencia y Análisis Asintótico

Evaluando en $s = j\omega$ (con $s^2 = -\omega^2$):

$$H(j\omega) = \frac{j\sqrt{2}\omega}{-\omega^2 + j\sqrt{2}\omega + 1} = \frac{j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$

#### A. Módulo Analítico $|H(j\omega)|$ y Ganancia en dB

$$|H(j\omega)| = \frac{\sqrt{0^2 + (\sqrt{2}\omega)^2}}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}} = \frac{\sqrt{2}\omega}{\sqrt{(1 - 2\omega^2 + \omega^4) + 2\omega^2}} = \frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}}$$

En decibeles:
$$|H(j\omega)|_{\text{dB}} = 20 \log_{10} \left( \frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}} \right) = 20 \log_{10}(\sqrt{2}\omega) - 10 \log_{10}(1 + \omega^4)$$
$$|H(j\omega)|_{\text{dB}} \approx 3{,}01 + 20 \log_{10}(\omega) - 10 \log_{10}(1 + \omega^4)$$

#### Análisis Asintótico de Módulo:
* **Bajas frecuencias ($\omega \to 0$, $\omega \ll 1$):**
  $$1 + \omega^4 \approx 1$$
  $$|H(j\omega)| \approx \sqrt{2}\omega$$
  $$|H(j\omega)|_{\text{dB}} \approx 20 \log_{10}(\sqrt{2}\omega) = 3{,}01 + 20 \log_{10}(\omega)$$
  *(Asíntota de baja frecuencia con pendiente de $+20\text{ dB/década}$ debido al cero en el origen $s=0$. La recta cruza $0\text{ dB}$ en $\omega = \frac{1}{\sqrt{2}} \approx 0{,}707\text{ rad/s}$).*

* **Altas frecuencias ($\omega \to \infty$, $\omega \gg 1$):**
  $$1 + \omega^4 \approx \omega^4 \implies \sqrt{1 + \omega^4} \approx \omega^2$$
  $$|H(j\omega)| \approx \frac{\sqrt{2}\omega}{\omega^2} = \frac{\sqrt{2}}{\omega}$$
  $$|H(j\omega)|_{\text{dB}} \approx 20 \log_{10}(\sqrt{2}) - 20 \log_{10}(\omega) \approx 3{,}01 - 20 \log_{10}(\omega)$$
  *(Asíntota de alta frecuencia con pendiente de $-20\text{ dB/década}$. La recta cruza $0\text{ dB}$ en $\omega = \sqrt{2} \approx 1{,}414\text{ rad/s}$).*

* **Frecuencia característica de resonancia ($\omega = \omega_0 = 1\text{ rad/s}$):**
  $$|H(j1)| = \frac{\sqrt{2} \cdot 1}{\sqrt{1 + 1^4}} = \frac{\sqrt{2}}{\sqrt{2}} = 1$$
  $$|H(j1)|_{\text{dB}} = 20 \log_{10}(1) = 0\text{ dB}$$
  *(En resonancia, el tanque paralelo entra en antirresonancia comportándose como un circuito abierto ideal, logrando ganancia unitaria y máxima de $0\text{ dB}$).*


#### B. Fase Analítica $\phi(\omega)$

$$\phi(\omega) = \arg\{j\sqrt{2}\omega\} - \arg\{(1 - \omega^2) + j\sqrt{2}\omega\}$$

Analizando el ángulo del numerador $N(j\omega) = j\sqrt{2}\omega$:
* Es un número imaginario puro positivo para todo $\omega > 0$:
  $$\arg\{N(j\omega)\} = +\frac{\pi}{2} = +90^\circ$$

Analizando el ángulo del denominador $D(j\omega) = (1 - \omega^2) + j\sqrt{2}\omega$:
* Si $\omega < 1$: primer cuadrante $\implies \arg\{D\} = \arctan\left( \frac{\sqrt{2}\omega}{1 - \omega^2} \right)$
* Si $\omega = 1$: semieje imaginario positivo $\implies \arg\{D\} = \frac{\pi}{2}$
* Si $\omega > 1$: segundo cuadrante $\implies \arg\{D\} = \pi - \arctan\left( \frac{\sqrt{2}\omega}{\omega^2 - 1} \right)$

#### Evaluación de Puntos Notables de Fase:
* **$\omega \to 0$:**
  $$\arg\{N\} = \frac{\pi}{2}, \quad \arg\{D\} \to \arctan(0) = 0$$
  $$\phi(0) = \frac{\pi}{2} - 0 = +\frac{\pi}{2}\text{ rad} = +90^\circ$$

* **$\omega = \omega_0 = 1\text{ rad/s}$:**
  $$\arg\{N\} = \frac{\pi}{2}, \quad \arg\{D\} = \frac{\pi}{2}$$
  $$\phi(1) = \frac{\pi}{2} - \frac{\pi}{2} = 0\text{ rad} = 0^\circ$$
  *(En la resonancia, la salida y la entrada están perfectamente en fase).*

* **$\omega \to \infty$:**
  $$\arg\{N\} = \frac{\pi}{2}, \quad \arg\{D\} \to \pi$$
  $$\phi(\infty) = \frac{\pi}{2} - \pi = -\frac{\pi}{2}\text{ rad} = -90^\circ$$


### 3. Clasificación y Ancho de Banda

Observando el diagrama de Bode y los puntos característicos, confirmamos que el circuito actúa como un **filtro pasa-banda (Band-Pass Filter) de 2.º orden**.

* **En bajas frecuencias ($\omega \to 0$):** El inductor $L$ se comporta como un cortocircuito a tierra ($Z_L \to 0$), derivando toda la señal a masa, por lo que la ganancia tiende a $-\infty\text{ dB}$ (módulo lineal nulo). La curva sube con una pendiente de $+20\text{ dB/década}$.
* **En la frecuencia de resonancia ($\omega_0 = 1\text{ rad/s}$):** El tanque paralelo $LC$ resuena en paralelo (impedancia infinita), de modo que no drena corriente a tierra y toda la tensión cae sobre la salida, alcanzando el pico máximo de transmisión de $0\text{ dB}$ (módulo $1$) y desfasaje nulo ($0^\circ$).
* **En altas frecuencias ($\omega \to \infty$):** El capacitor $C$ se comporta como un cortocircuito a tierra ($Z_C \to 0$), cortocircuitando nuevamente la salida a masa, con una atenuación asintótica de $-20\text{ dB/década}$.

#### Frecuencias de Corte a $-3\text{ dB}$ y Ancho de Banda
Las frecuencias de corte $\omega_c$ se definen en aquellos puntos donde la potencia cae a la mitad, es decir, donde la ganancia cae $3\text{ dB}$ respecto del máximo:
$$|H(j\omega)| = \frac{1}{\sqrt{2}}$$

Igualando el módulo:
$$\frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}} = \frac{1}{\sqrt{2}} \implies \frac{2\omega^2}{1 + \omega^4} = \frac{1}{2} \implies 4\omega^2 = 1 + \omega^4$$
$$\omega^4 - 4\omega^2 + 1 = 0$$

Resolviendo la ecuación bicuadrática para $\omega^2$:
$$\omega^2 = \frac{4 \pm \sqrt{16 - 4}}{2} = \frac{4 \pm \sqrt{12}}{2} = 2 \pm \sqrt{3}$$

Tomando las raíces positivas para las frecuencias de corte inferior ($\omega_{c1}$) y superior ($\omega_{c2}$):
$$\omega_{c1} = \sqrt{2 - \sqrt{3}} = \frac{\sqrt{6} - \sqrt{2}}{2} \approx 0{,}518\text{ rad/s}$$
$$\omega_{c2} = \sqrt{2 + \sqrt{3}} = \frac{\sqrt{6} + \sqrt{2}}{2} \approx 1{,}932\text{ rad/s}$$

Verificamos la simetría geométrica:
$$\omega_{c1} \cdot \omega_{c2} = \sqrt{(2 - \sqrt{3})(2 + \sqrt{3})} = \sqrt{4 - 3} = 1 = \omega_0^2$$

El ancho de banda ($\text{BW}$) del filtro resulta:
$$\text{BW} = \omega_{c2} - \omega_{c1} = \frac{\omega_0}{Q} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2} \approx 1{,}414\text{ rad/s}$$

$$
\text{BW} = [\omega_{c1}, \omega_{c2}] = [0{,}518, 1{,}932]\text{ rad/s}
$$

Por lo tanto, el filtro deja pasar las componentes frecuenciales comprendidas en el intervalo $[0{,}518, 1{,}932]\text{ rad/s}$, atenuando las frecuencias fuera de esta banda a razón de $\pm 20\text{ dB/década}$.
